# Medical PDF question-answering workflow

This notebook is a safe, credential-free overview of the repository pipeline. Configure credentials in a local `.env` file; never paste API keys into notebook cells or outputs.

Run `python store_index.py` to prepare the Pinecone index and `python app.py` to start the Flask app. The application reads `data/Medical_book.pdf`, splits its text, embeds the chunks, retrieves relevant passages, and asks a local Llama 2 model to draft a response.
